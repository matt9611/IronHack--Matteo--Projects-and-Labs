```python
list1 = [ "t-shirt", "mug", "hat", "book", "keychain" ]
inventory = {}
index=0

while index < len(list1):
    item = list1[index]
    inventory[item] = int(input(f"Insert quantity of {item}: "))
    index += 1 

print(inventory)
```

```python
customer_orders = set()
x=True 
while x==True:
    
    order= input("Write the product you want :")
    
    if order in list1:
        customer_orders.add(order)
    else:
        print("item not found")
        
    y=input("do you eant to continue yes/no?")
    if y == "no":
        x=False
        
print(customer_orders)
```

```python
for order in customer_orders:
    if order in inventory:
        inventory[order] -= 1  # Reduce the count of the item by 1

print(inventory)
```
