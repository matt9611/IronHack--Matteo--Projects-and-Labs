```python
list1 = [ "t-shirt", "mug", "hat", "book", "keychain" ]
inventory = {}

for item in (list1):
    inventory[item] = int(input(f" insert quantity of {item}"))

print(inventory)
    

```

```python

customer_orders = set()
x=1
while x<=3:
    
    order= input(f"{x}Write the product you want :")
    x=x+1
    if order in list1:
        customer_orders.add(order)
        
print(customer_orders)
    
    
```

```python
print("number of items ordered:",len(customer_orders))
```

```python
Percentage= (len(customer_orders)/5)*100
```

```python
Percentage
```

```python
numbers_items= len(customer_orders)
order_status = (numbers_items, Percentage)
order_status
```

```python
print(f"""Order statistics:
total products ordered:{order_status[0]}
Percentage of products ordered : {order_status[1]}""")
```
