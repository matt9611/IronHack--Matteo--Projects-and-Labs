```python
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_quantity = False
        while not valid_quantity:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
                valid_quantity = True
            except ValueError as error:
                print(f"Error: {error}")
        inventory[product] = quantity
    return inventory

initialize_inventory(["chair", "mug"])
```

```python
def total_price(c_orders):
    tot = 0  # Initialize total as zero

    # Iterate through each order in the list
    for i in c_orders:
        valid_quantity = False
        while not valid_quantity:
            try:
                price = int(input(f"Insert price for {i}: "))
                if price < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
                # this check is logically unnecessary now
                elif type(price) == str:
                    raise ValueError("Invalid quantity! Please enter a numerical value.")

                valid_quantity = True  # Set to True only when the price input is valid
            except ValueError as error:
                print(f"error: {error}")
                return None 

        tot += price  # Add the inputted price to the total
        
    return tot  # Return the total after all iterations are complete

# Example usage
orders = ["chair", "Phone"]
total = total_price(orders)
print(f"Total price: {total}")
```
