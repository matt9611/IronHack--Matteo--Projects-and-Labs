```python
import pandas as pd

url = 'https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis.csv'
df = pd.read_csv(url)
```

```python
df['Vehicle Type'].drop
```

```python
df = df.drop('Vehicle Type', axis=1)
```

```python
df.head()
```

```python
df.sample(10)
```

```python
VehicleClass_mode = df['Vehicle Class'].mode()[0]
df['Vehicle Class'] = df['Vehicle Class'].fillna(VehicleClass_mode)
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df['Response'] = df['Response'].fillna('No')
```

```python
df.isna().sum()
```

```python
pd.set_option('display.max_columns', None)
df.sample(20)
```

```python
state_mode = df['State'].mode()[0]
df['State'] = df['State'].fillna(state_mode)

# Fill specific numerical columns with zero if that makes sense
df['Months Since Last Claim'] = df['Months Since Last Claim'].fillna(0)
df['Number of Open Complaints'] = df['Number of Open Complaints'].fillna(0)

# Fill 'Vehicle Size' with its mode
vehicle_size_mode = df['Vehicle Size'].mode()[0]
df['Vehicle Size'] = df['Vehicle Size'].fillna(vehicle_size_mode)
```

```python
df.isna().sum()
```

```python
df.head()
```

```python
filtered_df = df[(df['Total Claim Amount'] < 1000) & (df['Response'] == 'Yes')]
```
