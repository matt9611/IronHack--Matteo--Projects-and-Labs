```python
url = 'https://raw.githubusercontent.com/data-bootcamp-v4/data/main/file1.csv'
df = pd.read_csv(url)
```

```python
df['gender'].unique()
```

```python
df['gender'] = df['gender'].replace({'Male':'M', 'Femal':'F', 'female':'F'   })



print(df.head())
```

```python
df['state'].unique()
```

```python
df['state'] = df['state'].replace({'AZ':'Arizona', 'WA':'Washinghton', 'Cali':'California'   })
```

```python
print(df.head())
```

```python
df['education'] = df['education'].replace({'Bachelors':'Bachelor'})
```

```python
df['education'].unique()
```

```python
ef=df.copy()
ef['customer_lifetime_value'] = df['customer_lifetime_value'].str.replace('%','')
```

```python
print(ef.head())
```

```python
ef['vehicle_class'] = ef['vehicle_class'].replace({'Sports Car':'Luxury','Luxury SUV':'Luxury', 'Luxury Car':'Luxury' })
```

```python
print(ef.sample(5))
```

```python
ef['customer_lifetime_value'] = pd.to_numeric(ef['customer_lifetime_value'], errors='coerce')
```

```python
print(ef.sample(5))
```

```python
most_frequent_state = df['state'].mode()[0]  # mode()[0] extracts the actual value

# Fill NaN values in 'state' column with the most frequent state
df['state'] = df['state'].fillna(most_frequent_state)
```
